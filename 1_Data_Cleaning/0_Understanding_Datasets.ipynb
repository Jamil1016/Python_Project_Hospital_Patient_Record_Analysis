```python
df_patients['BIRTHDATE'] = pd.to_datetime(df_patients['BIRTHDATE'])
df_patients['DEATHDATE'] = pd.to_datetime(df_patients['DEATHDATE'])

def cleaned(text):
    if pd.notna(text):
        return "".join(c for c in text if not(c.isdigit()))
    else: 
        return np.nan

df_patients['FIRST'] = df_patients['FIRST'].apply(cleaned)
df_patients['LAST'] = df_patients['LAST'].apply(cleaned)
df_patients['MAIDEN'] = df_patients['MAIDEN'].apply(cleaned)

df_patients
```
